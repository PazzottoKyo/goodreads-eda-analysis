```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style = "whitegrid", context = "notebook")

df_raw = pd.read_csv("../data/raw_data/books_colunas_relevantes.csv")

print(f"Dimensão: {df_raw.shape[0]} linhas X {df_raw.shape[1]} colunas\n")
print(df_raw.dtypes)
print("\nValores ausentes:\n", df_raw.isnull().sum())
print("\nDuplicatas:", df_raw.duplicated(subset = ["title", "authors"]).sum())

df_raw.head()
```

```python
def classificar_periodo(ano):
    if pd.isna(ano):
        return "Desconhecido"
    elif ano < 1900:
        return "Clássicos (Pré-1900)"
    elif ano < 1980:
        return "Século XX (1900-1979)"
    elif ano < 2000:
        return "Contemporâneo (1980-1999)"
    else:
        return "Século XXI (2000+)"
        

def padronizar_idioma(codigo):
    variacoes_ingles = ["eng", "en-us", "en-gb", "en-ca", "en"]
    if pd.isna(codigo):
        return "desconhecido"
    codigo_limpo = str(codigo).strip().lower()
    if codigo_limpo in variacoes_ingles:
        return "eng"
    return codigo_limpo
    

def verificar_anos_antigos(dataframe):
    antigos = dataframe[dataframe["original_publication_year"] <= 0]
    print(f"Obras com ano <= 0: {len(antigos)}")
    return antigos[["title", "authors", "original_publication_year"]].head()
```

```python
print(verificar_anos_antigos(df_raw))

df_clean = df_raw.dropna(subset=["original_publication_year"]).copy()
df_clean["original_publication_year"] = df_clean["original_publication_year"].astype(int)

df_clean["idioma_limpo"] = df_clean["language_code"].apply(padronizar_idioma)
df_clean["periodo"] = df_clean["original_publication_year"].apply(classificar_periodo)

df_clean["log_ratings"] = np.log10(df_clean["ratings_count"] + 1)
df_clean["log_reviews"] = np.log10(df_clean["work_text_reviews_count"] + 1)

df_clean.to_csv("../data/processed_data/books_clean.csv", index=False)
print(f"\nDimensão após limpeza: {df_clean.shape[0]} linhas X {df_clean.shape[1]} colunas")
```

```python
def resumir_variavel(serie, nome):
    return {
        "Variável": nome,
        "Média": round(serie.mean(), 3),
        "Mediana": round(serie.median(), 3),
        "Desvio-Padrão": round(serie.std(), 3),
        "Mínimo": round(serie.min(), 3),
        "Máximo": round(serie.max(), 3)
    }


tabela_descritiva = pd.DataFrame([
    resumir_variavel(df_clean["average_rating"], "Avaliação Média"),
    resumir_variavel(df_clean["ratings_count"], "Qtd. de Avaliações"),
    resumir_variavel(df_clean["work_text_reviews_count"], "Qtd. de Resenhas"),
    resumir_variavel(df_clean["original_publication_year"], "Ano de Publicação")
])

print(tabela_descritiva.to_string(index = False))
```

```python
plt.figure(figsize = (8, 4.5))
sns.histplot(df_clean["average_rating"], bins = 30, kde=True, color = "steelblue")
plt.axvline(df_clean["average_rating"].mean(), color = "red", linestyle = "--", label = f"Média: {df_clean['average_rating'].mean():.2f}")
plt.axvline(df_clean["average_rating"].median(), color = "darkgreen", linestyle = ":", label = f"Mediana: {df_clean['average_rating'].median():.2f}")
plt.title("Distribuição das Avaliações Médias no Goodreads")
plt.xlabel("Avaliação Média")
plt.ylabel("Quantidade de Livros")
plt.legend()
plt.tight_layout()
plt.savefig("../charts/01_histograma_avaliacoes.png", dpi = 300)
plt.show()
```

```python
mapa_periodos_en = {
    "Clássicos (Pré-1900)": "Classics (Pre-1900)",
    "Século XX (1900-1979)": "20th Century (1900-1979)",
    "Contemporâneo (1980-1999)": "Contemporary (1980-1999)",
    "Século XXI (2000+)": "21st Century (2000+)"
}

df_clean["period_en"] = df_clean["periodo"].map(mapa_periodos_en)
ordem_en = list(mapa_periodos_en.values())

print(df_clean.groupby("period_en")["average_rating"].agg(["count", "mean", "median", "std"]).loc[ordem_en].round(3))
```
