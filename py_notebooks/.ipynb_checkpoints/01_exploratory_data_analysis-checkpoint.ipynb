```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style = "whitegrid", context = "notebook")

df_raw = pd.read_csv("../data/raw_data/books_colunas_relevantes.csv")

print(f"Dimensão: {df_raw.shape[0]} linhas X {df_raw.shape[1]} colunas\n")
print(df_raw.dtypes)
print("\nValores ausentes:\n", df_raw.isnull().sum())
print("\nDuplicatas:", df_raw.duplicated(subset = ["title", "authors"]).sum())

df_raw.head()
```

```python
def classificar_periodo(ano):
    if pd.isna(ano):
        return "Desconhecido"
    elif ano < 1900:
        return "Clássicos (Pré-1900)"
    elif ano < 1980:
        return "Século XX (1900-1979)"
    elif ano < 2000:
        return "Contemporâneo (1980-1999)"
    else:
        return "Século XXI (2000+)"
        

def padronizar_idioma(codigo):
    variacoes_ingles = ["eng", "en-us", "en-gb", "en-ca", "en"]
    if pd.isna(codigo):
        return "desconhecido"
    codigo_limpo = str(codigo).strip().lower()
    if codigo_limpo in variacoes_ingles:
        return "eng"
    return codigo_limpo
    

def verificar_anos_antigos(dataframe):
    antigos = dataframe[dataframe["original_publication_year"] <= 0]
    print(f"Obras com ano <= 0: {len(antigos)}")
    return antigos[["title", "authors", "original_publication_year"]].head()
```

```python
print(verificar_anos_antigos(df_raw))

df_clean = df_raw.dropna(subset = ["original_publication_year"]).copy()
df_clean["original_publication_year"] = df_clean["original_publication_year"].astype(int)

df_clean["idioma_limpo"] = df_clean["language_code"].apply(padronizar_idioma)
df_clean["periodo"] = df_clean["original_publication_year"].apply(classificar_periodo)

df_clean["log_ratings"] = np.log10(df_clean["ratings_count"] + 1)
df_clean["log_reviews"] = np.log10(df_clean["work_text_reviews_count"] + 1)

df_clean.to_csv("../data/processed_data/books_clean.csv", index = False)
print(f"\nDimensão após limpeza: {df_clean.shape[0]} linhas X {df_clean.shape[1]} colunas")
```

```python
def resumir_variavel(serie, nome):
    return {
        "Variável": nome,
        "Média": round(serie.mean(), 3),
        "Mediana": round(serie.median(), 3),
        "Desvio-Padrão": round(serie.std(), 3),
        "Mínimo": round(serie.min(), 3),
        "Máximo": round(serie.max(), 3)
    }


tabela_descritiva = pd.DataFrame([
    resumir_variavel(df_clean["average_rating"], "Avaliação Média"),
    resumir_variavel(df_clean["ratings_count"], "Qtd. de Avaliações"),
    resumir_variavel(df_clean["work_text_reviews_count"], "Qtd. de Resenhas"),
    resumir_variavel(df_clean["original_publication_year"], "Ano de Publicação")
])

print(tabela_descritiva.to_string(index = False))
```

```python
plt.figure(figsize = (8, 4.5))
sns.histplot(df_clean["average_rating"], bins = 30, kde=True, color = "steelblue")
plt.axvline(df_clean["average_rating"].mean(), color = "red", linestyle = "--", label = f"Média: {df_clean['average_rating'].mean():.2f}")
plt.axvline(df_clean["average_rating"].median(), color = "darkgreen", linestyle = ":", label = f"Mediana: {df_clean['average_rating'].median():.2f}")
plt.title("Distribuição das Avaliações Médias no Goodreads")
plt.xlabel("Avaliação Média")
plt.ylabel("Quantidade de Livros")
plt.legend()
plt.tight_layout()
plt.savefig("../charts/01_histograma_avaliacoes.png", dpi = 300)
plt.show()
```

```python
ordem = [
    "Clássicos (Pré-1900)",
    "Século XX (1900-1979)",
    "Contemporâneo (1980-1999)",
    "Século XXI (2000+)"
]

tabela_periodo = (
    df_clean.groupby("periodo")["average_rating"]
    .agg(["count", "mean", "median", "std"])
    .loc[ordem]
    .rename(columns={
        "count": "Quantidade",
        "mean": "Média",
        "median": "Mediana",
        "std": "Desvio-Padrão"
    })
    .round(3)
)

tabela_periodo.index.name = "Período"
print(tabela_periodo)
```

```python
plt.figure(figsize = (9, 6.5))
sns.boxplot(data = df_clean, x = "periodo", y = "average_rating", order = ordem, hue = "periodo", palette = "Blues", legend = False, width = 0.5)
plt.title("Avaliação Média por Período de Publicação", pad = 15)
plt.xlabel("Período Histórico", labelpad = 10)
plt.ylabel("Avaliação Média", labelpad = 10)
plt.ylim(2.2, 5.0)
plt.xticks(rotation = 10)
plt.tight_layout()
plt.savefig("../charts/02_boxplot_periodo.png", dpi = 300)
plt.show()
```

```python
colunas = {
    "average_rating": "Avaliação Média",
    "ratings_count": "Qtd. Avaliações",
    "work_text_reviews_count": "Qtd. Resenhas",
    "log_ratings": "Log10(Avaliações)",
    "log_reviews": "Log10(Resenhas)"
}

print(df_clean[list(colunas.keys())].rename(columns = colunas).corr().round(3))
```

```python
df_idiomas_conhecidos = df_clean[df_clean["idioma_limpo"] != "desconhecido"]
top5_idiomas = df_idiomas_conhecidos["idioma_limpo"].value_counts().head(5).index.tolist()
df_top5 = df_idiomas_conhecidos[df_idiomas_conhecidos["idioma_limpo"].isin(top5_idiomas)]

tabela_idiomas = (
    df_top5.groupby("idioma_limpo")["average_rating"]
    .agg(["count", "mean", "median", "std"])
    .loc[top5_idiomas]
    .rename(columns = {
        "count": "Quantidade",
        "mean": "Média",
        "median": "Mediana",
        "std": "Desvio-Padrão"
    })
    .round(3)
)

tabela_idiomas.index.name = "Idioma"
print(tabela_idiomas)
```

```python
from scipy.stats import kruskal, spearmanr

#Teste de Kruskal-Wallis por Período Histórico
grupos_periodo = [grupo["average_rating"].values for _, grupo in df_clean.groupby("periodo")]
stat_periodo, p_periodo = kruskal(*grupos_periodo)
```

```python
#Teste de Kruskal-Wallis pelos Top 5 Idiomas
df_idiomas_conhecidos = df_clean[df_clean["idioma_limpo"] != "desconhecido"]
top5_idiomas = df_idiomas_conhecidos["idioma_limpo"].value_counts().head(5).index.tolist()
df_top5 = df_idiomas_conhecidos[df_idiomas_conhecidos["idioma_limpo"].isin(top5_idiomas)]

grupos_idioma = [grupo["average_rating"].values for _, grupo in df_top5.groupby("idioma_limpo")]
stat_idioma, p_idioma = kruskal(*grupos_idioma)
```

```python
#Correlação de Spearman (Popularidade e Resenhas vs. Avaliação Média)
rho_ratings, p_ratings = spearmanr(df_clean["ratings_count"], df_clean["average_rating"])
rho_reviews, p_reviews = spearmanr(df_clean["work_text_reviews_count"], df_clean["average_rating"])
```

```python
resultados_testes = pd.DataFrame({
    "Hipótese / Comparação": [
        "Avaliação Média entre Períodos Históricos (Kruskal-Wallis)",
        "Avaliação Média entre Top 5 Idiomas (Kruskal-Wallis)",
        "Popularidade (Qtd. Avaliações) vs. Avaliação (Spearman)",
        "Engajamento (Qtd. Resenhas) vs. Avaliação (Spearman)"
    ],
    "Estatística": [
        round(stat_periodo, 3),
        round(stat_idioma, 3),
        round(rho_ratings, 3),
        round(rho_reviews, 3)
    ],
    "Valor-p (p-value)": [
        f"{p_periodo:.2e}",
        f"{p_idioma:.2e}",
        f"{p_ratings:.2e}",
        f"{p_reviews:.2e}"
    ],
    "Significativo (α = 0.05)?": [
        "Sim" if p_periodo < 0.05 else "Não",
        "Sim" if p_idioma < 0.05 else "Não",
        "Sim" if p_ratings < 0.05 else "Não",
        "Sim" if p_reviews < 0.05 else "Não"
    ]
})

print(resultados_testes.to_string(index = False))
```

## Validação Estatística (Testes de Hipóteses)

A aplicação de testes não paramétricos (adequados devido à assimetria observada nas distribuições de notas e engajamento) confirma matematicamente os padrões identificados na etapa visual ($N = 9.979$, $\alpha = 0,05$):


### 1. Comparação entre Grupos (Teste de Kruskal-Wallis)

* **Períodos Históricos ($H = 93,206$, $p = 4,48 \times 10^{-20}$):** Existe uma diferença estatisticamente significativa nas avaliações médias entre as épocas de publicação. Esse resultado corrobora o padrão observado no gráfico de caixa (*boxplot*), indicando o efeito do **viés de sobrevivência**: obras mais antigas e clássicas que permanecem relevantes no catálogo tendem a manter notas superiores às de lançamentos contemporâneos.
* **Top 5 Idiomas ($H = 22,476$, $p = 1,61 \times 10^{-4}$):** Há diferença estatisticamente significativa na distribuição das avaliações entre os cinco idiomas mais frequentes, embora a amostra seja fortemente concentrada em obras de língua inglesa.


### 2. Análise de Correlação Monotônica (Spearman)

* **Popularidade vs. Avaliação Média ($\rho = 0,079$, $p = 2,53 \times 10^{-15}$):** Ilustra a distinção clássica entre **significância estatística** e **relevância prática** em grandes amostras. Embora o valor-$p$ seja extremamente baixo devido ao tamanho do conjunto de dados, o coeficiente de Spearman ($\rho = 0,079$) revela uma correlação positiva desprezível — ou seja, a alta popularidade de um livro praticamente não eleva sua nota média.
* **Engajamento Textual vs. Avaliação Média ($\rho = -0,149$, $p = 7,80 \times 10^{-51}$):** Observa-se uma correlação negativa fraca, porém estatisticamente robusta. Livros com maior volume de resenhas escritas tendem a apresentar avaliações médias ligeiramente inferiores, sugerindo que obras mais polêmicas ou divisivas estimulam maior debate textual entre os leitores do que consensos unânimes.


> **Síntese para o Estudo:** A qualidade percebida (nota média) no catálogo do Goodreads é influenciada pelo recorte temporal e linguístico, mas mantém-se praticamente independente do volume total de vendas/popularidade e apresenta leve tendência de queda quando o volume de debates escritos aumenta.

```python
top10_livros = (
    df_clean[["title", "authors", "original_publication_year", "average_rating", "ratings_count"]]
    .sort_values(by = "ratings_count", ascending = False)
    .head(10)
    .rename(columns = {
        "title": "Título",
        "authors": "Autor(es)",
        "original_publication_year": "Ano",
        "average_rating": "Avaliação Média",
        "ratings_count": "Qtd. Avaliações"
    })
)

print("Top 10 livros mais avaliados")
print(top10_livros.to_string(index = False))
print("\n" + "="*60 + "\n")
```

```python
top10_autores = (
    df_clean.groupby("authors")
    .agg(
        Qtd_Livros = ("title", "count"),
        Avaliacao_Media = ("average_rating", "mean"),
        Total_Avaliacoes = ("ratings_count", "sum")
    )
    .sort_values(by = "Total_Avaliacoes", ascending = False)
    .head(10)
    .round({"Avaliacao_Media": 2})
)

print("Top 10 autores por volume de avaliações")
print(top10_autores)
```

```python
fig, axes = plt.subplots(1, 2, figsize = (15, 6))
titulos_curtos = top10_livros["Título"].apply(lambda x: x[:32] + "..." if len(x) > 35 else x)

sns.barplot(
    x = top10_livros["Qtd. Avaliações"] / 1e6,
    y = titulos_curtos,
    hue = titulos_curtos,
    palette = "viridis",
    legend = False,
    ax = axes[0]
)

axes[0].set_title("Top 10 Livros Mais Avaliados", pad = 12)
axes[0].set_xlabel("Avaliações (em milhões)", labelpad = 8)
axes[0].set_ylabel("")

sns.barplot(
    x = top10_autores["Total_Avaliacoes"] / 1e6,
    y = top10_autores.index,
    hue = top10_autores.index,
    palette = "magma",
    legend = False,
    ax = axes[1]
)
axes[1].set_title("Top 10 Autores Mais Avaliados no Catálogo", pad = 12)
axes[1].set_xlabel("Total de Avaliações Acumuladas (em milhões)", labelpad = 8)
axes[1].set_ylabel("")

plt.tight_layout()
plt.savefig("../charts/06_top_livros_autores.png", dpi = 300)
plt.show()
```

## Conclusões Gerais da Análise Exploratória (EDA)

A investigação empírica da base higienizada do Goodreads ($N = 9.979$ obras válidas) permite sintetizar os seguintes achados estruturais sobre o comportamento das avaliações literárias:


### 1. Dinâmica de Distribuição e Popularidade

* **Assimetria Extrema (Cauda Longa):** Tanto o volume de avaliações (`ratings_count`) quanto o número de resenhas textuais (`work_text_reviews_count`) apresentam forte concentração em poucos títulos de massa e uma longa cauda de obras de nicho, exigindo transformações logarítmicas ($\log_{10}$) e abordagens estatísticas não paramétricas.
* **Independência entre Popularidade e Nota Média:** A correlação monotônica de Spearman entre o total de avaliações e a nota média é praticamente nula ($\rho = 0,079$, $p = 2,53 \times 10^{-15}$), comprovando que o alcance comercial de um livro não se traduz em avaliações superiores.


### 2. Fatores Temporais, Linguísticos e de Engajamento

1. **Viés de Sobrevivência Temporal:** O teste de Kruskal-Wallis ($H = 93,206$, $p = 4,48 \times 10^{-20}$) confirma diferenças estatisticamente significativas entre os períodos históricos. Obras publicadas antes de 1900 (*Clássicos*) mantêm medianas de avaliação superiores às dos lançamentos contemporâneos, reflexo da filtragem histórica que preserva apenas obras consagradas.
2. **Efeito de Polarização nas Resenhas:** Observa-se uma correlação negativa fraca, porém significativa, entre o volume de resenhas escritas e a nota média ($\rho = -0,149$, $p = 7,80 \times 10^{-51}$), sugerindo que obras mais divisivas ou polêmicas estimulam maior debate textual do que livros de consenso unânime.
3. **Concentração Linguística:** Embora existam variações estatisticamente significativas nas notas entre os cinco idiomas mais frequentes ($H = 22,476$, $p = 1,61 \times 10^{-4}$), o catálogo exibe forte hegemonia de publicações em língua inglesa.

> **Síntese Final:** O ecossistema do Goodreads caracteriza-se por uma qualidade percebida relativamente estável em torno de 4,0 estrelas, na qual o recorte histórico e o grau de controvérsia textual exercem maior influência sobre a nota média do que a popularidade bruta da obra.
